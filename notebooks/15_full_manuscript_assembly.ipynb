## 02. Imports and paths

```python
from pathlib import Path
import pandas as pd
from datetime import datetime

PROJECT_DIR = Path('/content/drive/MyDrive/PPMI_PD_Progression')
NB14_DIR = PROJECT_DIR / 'outputs' / 'notebook_14_manuscript_drafting_package'
OUT_DIR = PROJECT_DIR / 'outputs' / 'notebook_15_full_manuscript_assembly'
OUT_DIR.mkdir(parents=True, exist_ok=True)

print('NB14_DIR:', NB14_DIR, '| exists:', NB14_DIR.exists())
print('OUT_DIR:', OUT_DIR)
if not NB14_DIR.exists():
    raise FileNotFoundError(f'Notebook 14 output folder not found: {NB14_DIR}')
```

## 03. Load Notebook 14 drafting files

```python
required_files = {
    'title_key_messages': '01_title_and_key_messages.md',
    'abstract': '02_structured_abstract_draft.md',
    'methods': '03_methods_draft.md',
    'results': '04_results_draft.md',
    'captions': '05_table_captions_and_figure_legends.md',
    'limitations': '06_limitations_and_future_work.md',
    'data_ethics_ack': '07_data_availability_ethics_acknowledgement.md',
    'reporting_checklist': '08_reporting_checklist.csv',
    'submission_checklist': '09_submission_readiness_checklist.csv',
    'summary': '10_notebook_14_summary_report.txt',
}

loaded = {}
rows = []
for key, filename in required_files.items():
    path = NB14_DIR / filename
    exists = path.exists()
    rows.append({'key': key, 'filename': filename, 'exists': exists, 'size_bytes': path.stat().st_size if exists else None})
    if exists and filename.endswith('.md'):
        loaded[key] = path.read_text(encoding='utf-8', errors='replace')
    elif exists and filename.endswith('.txt'):
        loaded[key] = path.read_text(encoding='utf-8', errors='replace')

section_inventory = pd.DataFrame(rows)
section_inventory.to_csv(OUT_DIR / '02_manuscript_section_inventory.csv', index=False)
section_inventory
```

## 04. Build full manuscript draft

This cell uses a conservative manuscript structure and inserts placeholders where journal-specific items are still required.

```python
recommended_title = 'A Reproducible Machine Learning Benchmark for Predicting Rapid Motor Progression in Parkinson’s Disease Using PPMI Data'

abstract = loaded.get('abstract', '').replace('# Structured Abstract Draft', '').strip()
methods = loaded.get('methods', '').replace('# Methods Draft', '').strip()
results = loaded.get('results', '').replace('# Results Draft', '').strip()
limitations = loaded.get('limitations', '').replace('# Limitations and Future Work', '').strip()
data_ethics_ack = loaded.get('data_ethics_ack', '').replace('# Data Availability, Ethics, and Acknowledgement Draft', '').strip()
captions = loaded.get('captions', '').replace('# Table Captions and Figure Legends', '').strip()

manuscript = f"""# {recommended_title}

## Author information
[Insert author names, affiliations, corresponding author details]

## Abstract

{abstract}

## Keywords
Parkinson’s disease; PPMI; machine learning; motor progression; MDS-UPDRS; DaTSCAN; alpha-synuclein seed amplification assay; reproducible research

## Introduction
Parkinson’s disease progression is heterogeneous, and early identification of individuals at higher risk of rapid motor worsening remains an important research goal. Longitudinal public or controlled-access cohorts provide an opportunity to evaluate reproducible machine learning workflows for progression prediction while transparently reporting limitations.

The present study was designed as a reproducible benchmark rather than a clinical deployment model. We evaluated whether baseline clinical variables, DaTSCAN/SBR imaging-derived features, and SAA biomarker features from PPMI could predict rapid motor progression defined using annualized change in MDS-UPDRS Part III.

## Methods

{methods}

## Results

{results}

## Discussion
This reproducible benchmark showed that predicting rapid motor progression in Parkinson’s disease remains challenging using the evaluated baseline feature sets. Although DaTSCAN/SBR and SAA biomarker features provided some metric-specific changes, incremental gains were limited or inconsistent. The findings support transparent benchmark reporting rather than clinical deployment claims.

The results also emphasize the importance of preserving simple clinical-only models as primary comparators when evaluating multimodal machine learning pipelines. The modest discrimination observed here suggests that additional longitudinal, digital, imaging, or molecular predictors may be necessary to substantially improve prediction of rapid motor progression.

## Limitations and Future Work

{limitations}

## Data Availability, Ethics, and Acknowledgements

{data_ethics_ack}

## Tables and Figure Legends

{captions}

## References
[Insert references. Include PPMI protocol/citation instructions, MDS-UPDRS references, DaTSCAN/SBR references, SAA biomarker references, and machine-learning reporting guideline references.]
"""

(OUT_DIR / '01_full_manuscript_draft.md').write_text(manuscript, encoding='utf-8')
print('Saved:', OUT_DIR / '01_full_manuscript_draft.md')
print('Characters:', len(manuscript))
```
